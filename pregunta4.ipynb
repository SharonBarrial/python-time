```python
from collections import defaultdict

#leer archivo
with open("ayuda_maricielo.txt", "r") as f:
  n, m = map(int, f.readline().split())
  G=defaultdict(list) #para valores que aún no existen
  indegree=[0]*n
  while u!=0 and v!=0:
    for n in range(m):
      u, v =map(int, f.readline().split())
      G[u].append(v)
      u, v=map(int, f.readline().split())
      G[u].append(v)
      indegree[v]+=1

#aplicar orden topológico
def topological_sort(nodos: dict[str, list[str]]) -> list[str]:
  result=[]
  pq= [i: {} for i in range(n) if indegree[i]==0]
  while pq:
    u = pq.pop(0)
    result.append(u)
    for v in G[u]:
      indegree[v] -=1
      if indegree[v] ==0:
        pq.append(v)


  print("Lista de tareas de Maricielo: ")
  print(result)

```
