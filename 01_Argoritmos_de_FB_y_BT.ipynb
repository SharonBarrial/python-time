## Ordenamiento por Seleccion (Selection Sort)

```python
def SelectionSort(lista):
  n = len(lista)
  for i in range(n-1):
    min_pos = i
    for j in range(i+1, n):
      if lista[j] < lista[min_pos]:
        min_pos = j

    if min_pos != i:
      #hacemos la permutacion
      lista[min_pos], lista[i] = lista[i], lista[min_pos]


```
