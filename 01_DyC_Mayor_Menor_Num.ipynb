## Divide y Venceras: Mayor y Menor numero en un arreglo no ordenado

```python
import random as r
```

```python
lista = [r.randint(100,500) for _ in range(10)]
print(lista)
```

## Funcion **maximo**

```python
#Recibe los siguientes parametros
#a es el arreglo de enteros no ordenado
#i es posicion del primer elemento del arreglo
#j es posicion del ultimo elemento del arreglo
#i == j seria un solo item???

def maximo(a,i,j):
  if i == j:
    #Se trata de un arreglo con un unico item
    return a[i]
  else:
    #(1) Dividir
    med = (i+j) // 2 #Solo devolvemos la parte entera

    #(2) Conquistar, significa la llamada(s) recursiva(s)
    #Maximo de la parte izq del arreglo
    maxi = maximo(a,i,med)
    #Maximo de la parte derecha del arreglo
    maxd = maximo(a,med+1,j)

    #(3) Combinar
    if maxi > maxd:
      return maxi
    else:
      return maxd
```

```python
max = maximo(lista,0,len(lista)-1)
print(max)
```

## Funcion **minimo**

```python
def minimo(a,i,j):
  if i == j:
    #Se trata de un arreglo con un unico item
    return a[i]
  else:
    #(1) Dividir
    med = (i+j) // 2 #Solo devolvemos la parte entera

    #(2) Conquistar, significa la llamada(s) recursiva(s)
    #Maximo de la parte izq del arreglo
    mini = minimo(a,i,med)
    #Maximo de la parte derecha del arreglo
    mind = minimo(a,med+1,j)

    #(3) Combinar
    if mini < mind:
      return mini
    else:
      return mind
```

```python
min = minimo(lista,0,len(lista)-1)
print(min)
```
