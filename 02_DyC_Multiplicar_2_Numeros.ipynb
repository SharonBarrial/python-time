## Divide y Venceras: Multiplicacion de dos numeros grandes

```python
#Parametros
# a y b son numeros enteros mayores de n digitos
def mult(a,b,n):
  if n == 1:
    return a*b
  else:
    #(1) Divide
    #Parte derecha e izquierda de a
    ai = a // 10**(n//2)
    ad = a % 10**(n//2)
    #Parte derecha e izquierda de b
    bi = b // 10**(n//2)
    bd = b % 10**(n//2)

    #(2) Conquista
    #Llamadas recursivas a la funcion mult
    #1ro: Multiplicamos las partes izq de cada numero
    z1 = mult(ai,bi, n//2) * 10**n
    #2do: 
    #(a)Multiplicamos la parte izq de a con la derecha de b
    #(b)Multiplicamos la parte der de a con la izquierda de b
    #Sumar (a) + (b) y el resultado lo multiplicamos por 10**(n//2)
    z2 = (mult(ai,bd, n//2) + mult(ad,bi, n//2)) * 10**(n//2)
    z3 = mult(ad,bd, n//2)

    #(3) Combinar
    return z1 + z2 + z3
```

```python
a = 98765678
b = 24680135

resultado = mult(a,b,8)
print(resultado)
```
