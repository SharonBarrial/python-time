```python
def limit_parties(presupuesto, parties):
    n = len(parties)
    dp = [[0] * (presupuesto + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        price, diversion = parties[i - 1]
        for j in range(1, presupuesto + 1):
            if price > j:
                dp[i][j] = dp[i - 1][j]
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i - 1][j - price] + diversion)

    sum_tarifas = 0
    sum_diversion = dp[n][presupuesto]
    j = presupuesto
    for i in range(n, 0, -1):
        if dp[i][j] != dp[i - 1][j]:
            price, _ = parties[i - 1]
            sum_tarifas += price
            j -= price

    return sum_tarifas, sum_diversion
```

```python
#Muestra 1
presupuesto = 50
parties = [(12, 3),
           (15, 8),
           (16, 9),
           (16, 6),
           (10, 2),
           (21, 9),
           (18, 4),
           (12, 4),
           (17, 8),
          (18, 9)]

tarifas, diversion = limit_parties(presupuesto, parties)
print(f'Suma de tarifas', tarifas, f'suma de diversión', diversion)
```

```python
#Muetsra 2
presupuesto = 50
parties = [(13, 8),
           (19, 10),
           (16, 8),
           (12, 9),
           (10, 2),
           (12, 8),
           (13, 5),
           (15, 5),
           (11, 7),
          (16, 2)]

tarifas, diversion = limit_parties(presupuesto, parties)
print(tarifas, diversion)
```
