```python
import math as mt

def calcular_distancia(farol1, farol2):
    a1, b1 = farol1
    a2, b2 = farol2
    return mt.sqrt((a2 - a1)**2 + (b2 - b1)**2)

def calcular_cable_utilizado(red):
    cable_utilizado = 0
    for i in range(1, len(red)):
        cable_utilizado += calcular_distancia(red[i-1], red[i])
    return cable_utilizado

def nueva_propuesta(faroles):
    n = len(faroles)
    dp = [0] * n
    dp[0] = 0
    dp[1] = calcular_distancia(faroles[0], faroles[1])

    for i in range(2, n):
        dp[i] = min(dp[i-1] + calcular_distancia(faroles[i-1], faroles[i]), dp[i-2] + calcular_distancia(faroles[i-2], faroles[i]))

    nueva_propuesta = []
    i = n - 1
    while i > 0:
        if dp[i] == dp[i-1] + calcular_distancia(faroles[i-1], faroles[i]):
            nueva_propuesta.append(faroles[i])
            i -= 1
        else:
            nueva_propuesta.append(faroles[i])
            i -= 2
    nueva_propuesta.append(faroles[0])
    nueva_propuesta.reverse()

    return nueva_propuesta, dp[n-1]
```

```python
#Figura 1
propuesta_inicial = [(0, 0),
                    (1, 3),
                    (3, 1),
                    (2, 2),
                    (3, 5),
                    (1, 7),
                    (8, 6),
                    (5, 3),
                    (8, 1),
                    (8, 4),
                    (10, 3),
                    (12, 1),
                    (13, 3),
                    (12, 5),
                    (12, 7),]
cable_utilizado_inicial = calcular_cable_utilizado(propuesta_inicial)

print("Propuesta inicial:")
print("Faroles:", propuesta_inicial)
print("Cable utilizado:", cable_utilizado_inicial)
```

```python
#Nueva propuesta a través de DP
nueva_propuesta, cable_utilizado_nueva = nueva_propuesta(propuesta_inicial)
print("Nueva propuesta:")
print("Faroles:", nueva_propuesta)
print("Cable utilizado:", cable_utilizado_nueva)
```

```python
import matplotlib.pyplot as plt

def visualizar_grafo(faroles):
    x = [farol[0] for farol in faroles]
    y = [farol[1] for farol in faroles]

    plt.figure()
    plt.scatter(x, y, color='blue')
    for i, farol in enumerate(faroles):
        plt.annotate(f'({farol[0]}, {farol[1]})', (farol[0], farol[1]), textcoords="offset points", xytext=(0,5), ha='center')
        if i > 0:
            plt.plot([faroles[i-1][0], farol[0]], [faroles[i-1][1], farol[1]], color='red')

    plt.xlabel('Coordenada x')
    plt.ylabel('Coordenada y')
    plt.title('Red de Faroles')
    plt.grid(True)
    plt.show()
```

```python
visualizar_grafo(propuesta_inicial)
```

```python
visualizar_grafo(nueva_propuesta)
```
