## Algoritmo de Backtacking - NReinas

```python
import numpy as np # trabajar con matrices o arrays
import matplotlib.pyplot as plt # graficos en Python
```

```python
tablero = [4,7,2,6,1,0,3,5]

for fil, col in enumerate(tablero,0):
  print(f"Hay una reina en la fila {fil} columna {col}")
```

```python
#Funcion draw para dibujar un tablero

def draw(tablero):
  n = len(tablero)
  #np.zeros devuelve un array del tamaño y tipo indicados inicializando sus valores con ceros
  b = np.zeros((n, n, 3), dtype=int)
  #print(b)
  b += [255, 128, 80]

  b[::2, ::2] = [255, 225, 120]
  b[1::2, 1::2] = [255, 225, 120]

  _, ax = plt.subplots()
  ax.imshow(b)
  for fil, col in enumerate(tablero):
    ax.text(col, fil, u"\u265b", fontsize=200/n, va="center", ha="center")
  ax.set(xticks=[], yticks=[])
  '''
  La función set(xticks=[], yticks=[]) toma un objeto de lista como argumento. Los elementos de la lista indican las posiciones en la acción correspondiente donde se mostrarán las marcas. 
  Este método marcará los puntos de datos en las posiciones dadas con marcas.
  '''
```

```python
draw(tablero)
```

```python
def validar(tablero, fil, col):
  for fil_i in range(fil):
    col_i = tablero[fil_i]
    delta = fil - fil_i #fila anterior
    if col in (col_i, col_i + delta, col_i-delta): #posiciones no validas de las coronas en la columna
      return False

  return True
```

```python
tablero1 = [1,3,-1,-1]
```

```python
draw(tablero1)
```

```python
# Funcion assert
"""
Devuelve el resultado de una condicion verificando que sea verdad y si no lo es lanza una excepcion o error.
"""

assert validar(tablero1, 2,2) == False
```

```python
#Generalizar la funcion para las N reinas

def nreinas(tablero, fil): #fil es la fila inicial del tablero
  n = len(tablero)
  if fil == n:
    #dibujar el tablero cuando se evalua la ultima fila
    draw(tablero)
  else:
    for col in range(n):
      #print(fil)
      if validar(tablero, fil, col):
        tablero[fil] = col
        #llamada recursiva para evaluar la siguiente fila del tablero hasta fil == n
        nreinas(tablero, fil+1) 
```

```python
#Ejecucion de funcion nreinas

n = 4
#tablero vacio
tablero2 = [-1]*n
#print (tablero2)
```

```python
nreinas(tablero2,0)
```

```python
nreinas([-1]*8,0)
```
