Generalizamos a traves de una funcion

```python
def stringmatch(text,p): # recibe texto y cadena
  n = len(text)
  l = len(p)
  resultado = []

  for i in range(n - l +1 ):
    if p == text[i:i+l]:
      resultado.append(i)

  return resultado
```

```python
stringmatch("upc complejidad algoritmica upc", "upc")
```

## Uso de la libreria **re** 

Esta libreria **re** se utiliza para encontrar patrones en el texto.

La función **search()** toma el patrón y el texto para escanear, y devuelve un objeto Match cuando se encuentra el patrón. Si el patrón no se encuentra, search() devuelve None.

Cada objeto Match contiene información sobre la naturaleza de la coincidencia, incluida la cadena de entrada original, la expresión regular utilizada, y la ubicación dentro de la cadena original donde el patrón ocurre.

```python
import re

texto = 'cursocursocursocurso'
count = 0
print(re.findall(r'so', texto))
print(len(re.findall(r'so', texto)))
res = re.search(r'so', texto)
print(res.group())

for match in re.finditer(r'so', texto):
    count += 1
    print("match", count, match.group(), "start index", match.start(), "End index", match.end())
```
